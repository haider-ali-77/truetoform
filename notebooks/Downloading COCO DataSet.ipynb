# Preparation for Dataset

```python
with open('/home/usman/fiftyone/coco-2017/raw/person_keypoints_train2017.json', 'r') as file:
    keypoints = json.load(file)
```

```python
keypoints.keys()
```

```python
images_df = pd.DataFrame(keypoints['images'])
images_df
```

```python
annotations_df = pd.DataFrame(keypoints['annotations'])
annotations_df.head()
```

```python
annotations_df = annotations_df[annotations_df['iscrowd'] == 0]
```

```python
keypoints_df = pd.DataFrame(annotations_df['keypoints'].to_list(), index=annotations_df.index)
```

```python
legs_kp_confidence_col = (np.array([11,12,15,16])*3+3)-1
```

```python
legs_kp_confidence_col
```

```python
annotations_df = annotations_df[(keypoints_df[35] > 0) & (keypoints_df[38] > 0) & (keypoints_df[47] > 0) & (keypoints_df[50] > 0)]
```

```python
annotations_df['image_id'].unique().shape
```

```python
dataset = foz.load_zoo_dataset(
    "coco-2017",
    split="train",
    label_types=["segmentation"],
    classes=["person"],
    image_ids = list(annotations_df['image_id'].unique()),
    dataset_dir='/home/usman/datasets/coco'
  )

dataset.name = "coco-2017-train-examples"
dataset.persistent = True
session = fo.launch_app(dataset)
```
